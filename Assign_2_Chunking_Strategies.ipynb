```python

sample_document = """
Section 1: Introduction to Artificial Intelligence

Artificial Intelligence, commonly known as AI, is a field of computer science
that focuses on creating machines and software capable of performing tasks
that normally require human intelligence. These tasks include understanding
language, recognizing images, making decisions, solving problems, and learning
from experience.

AI systems can process large amounts of data and identify patterns that may
not be obvious to humans. Modern AI applications are used in healthcare,
finance, education, transportation, customer service, and many other industries.


Section 2: Machine Learning

Machine Learning is a subset of Artificial Intelligence. Instead of explicitly
programming a computer with rules for every possible situation, machine
learning allows a system to learn patterns from data.

There are three major types of machine learning: supervised learning,
unsupervised learning, and reinforcement learning. Supervised learning uses
labeled data to train a model. Unsupervised learning attempts to discover
patterns or structures in data without predefined labels. Reinforcement
learning allows an agent to learn by interacting with an environment and
receiving rewards or penalties.


Section 3: Large Language Models

Large Language Models, or LLMs, are AI models designed to understand and
generate human language. They are trained using very large collections of
text data and can perform tasks such as answering questions, summarizing
documents, translating languages, and generating code.

LLMs work by predicting tokens based on the context provided to them.
Modern models can contain billions of parameters and can process complex
instructions. However, an LLM does not automatically know information from
your private documents unless that information is provided to it.


Section 4: Retrieval-Augmented Generation

Retrieval-Augmented Generation, commonly called RAG, combines information
retrieval with text generation. Instead of asking an LLM to answer a question
using only its existing knowledge, a RAG system first retrieves relevant
information from a collection of documents.

The retrieved documents are then provided to the language model as context.
The model uses this context to generate an answer. RAG is commonly used for
chatbots, company knowledge bases, document question-answering systems,
and research assistants.


Section 5: Document Chunking

Document chunking is the process of dividing a large document into smaller
pieces called chunks. Chunking is important because language models have
limitations on how much text they can process at once.

A good chunk should contain enough information to preserve meaning while
remaining small enough for efficient processing. Chunk size and chunk overlap
are two important parameters when designing a chunking strategy.

For example, a chunk size of 600 characters with an overlap of 150 characters
means that neighboring chunks can share some information. This overlap helps
prevent important context from being lost at chunk boundaries.


Section 6: Vector Embeddings

Vector embeddings represent text as numerical vectors. Text with similar
meaning tends to have vectors that are close to each other in vector space.

Embeddings are useful for semantic search because a system can compare the
vector representation of a user's question with vectors representing stored
documents.

For example, a user might search for "How does an AI model learn from data?"
while a document contains the phrase "Machine learning algorithms identify
patterns from training data." Embeddings can help identify that these two
pieces of text have related meanings.
"""
print(f"Document length: {len(sample_document)} characters, "
      f"{len(sample_document.split())} words, "
      f"{sample_document.count(chr(10)+chr(10))} paragraph breaks.")
```

```python
import re
from langchain_core.documents import Document

section_pattern = r"(?m)^Section\s+\d+\s*:\s*.*$"

matches = list(re.finditer(section_pattern, sample_document))

structure_chunks = []

for i, match in enumerate(matches):

    section_title = match.group().strip()

    start = match.start()
    end = matches[i + 1].start() if i + 1 < len(matches) else len(sample_document)

    section_text = sample_document[start:end].strip()

    structure_chunks.append(
        Document(
            page_content=section_text,
            metadata={
                "section": section_title
            }
        )
    )

print(f"Produced {len(structure_chunks)} chunks\n")

for i, doc in enumerate(structure_chunks):
    print(
        f"--- Chunk {i+1} | "
        f"metadata: {doc.metadata} "
        f"({len(doc.page_content)} chars) ---"
    )
    print(doc.page_content)
    print()
```
