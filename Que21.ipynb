```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['figure.dpi'] = 110
pd.set_option('display.max_colwidth', 40)
print("Environment ready ✅")
```

```python
backlog = pd.DataFrame([
    {
        "name": "selef-serving belling dashboard",
        "user_business_value": 8,
        "time_criticality": 3,
        "risk_reduction": 4,
        "job_size": 5,
    },
      {
        "name": "fix-rare-edge-case-crash",
        "user_business_value": 2,
        "time_criticality": 9,
        "risk_reduction": 6,
        "job_size": 1,
    },
      {
        "name": "new on boarding redesign flow",
        "user_business_value": 9,
        "time_criticality": 6,
        "risk_reduction": 4,
        "job_size": 8,
    },
      {
        "name": "API rate limit increase",
        "user_business_value": 5,
        "time_criticality": 3,
        "risk_reduction": 2,
        "job_size": 2,
    },






])
backlog
```

```python
# 3. Compute WSJF
backlog["cost_of_delay"] = (
    backlog["user_business_value"] + backlog["time_criticality"] + backlog["risk_reduction"]
)
backlog["wsjf"] = (backlog["cost_of_delay"] / backlog["job_size"]).round(2)



ranked_wsjf = backlog.sort_values("wsjf", ascending=False).reset_index(drop=True)
ranked_wsjf.index += 1
ranked_wsjf[[ "wsjf"]]
```
